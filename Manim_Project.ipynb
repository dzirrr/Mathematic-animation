# Animasi Eid Mubarak

```python
from manim import *
config.media_width = "70%"
class Eid_Mubarak(Scene):
  def construct(self):
    #Watermark
    rect = RoundedRectangle(corner_radius=0.3, height=1.5, width=1.5, color=PINK)
    obj = Circle(radius=0.4, color=PINK).next_to(rect, DOWN, buff=-1.15)
    dot = Dot(color=PINK).scale(0.6).next_to(obj, buff=0.15).shift(UP*0.4)
    ig = VGroup(rect, obj, dot).scale(0.5).shift(LEFT*2)
    my_id = Text("@na_dzirrr", font_size=27).next_to(ig, buff=0.1)
    self.play(FadeIn(my_id), Create(ig), run_time=0.5)
    wm = VGroup(ig, my_id)
    self.play(wm.animate.scale(0.4).move_to(DOWN*3))
    self.wait(0.3)

    #Selesaikan Soal Berikut
    teks = Text('Sederhanakan Persamaan Berikut').scale(0.3).set_color(color='#F38181').shift(UP*2.5)
    self.play(FadeIn(teks), run_time=0.8)
    self.wait()

    #Integral Eid Mubarak
    id1 = MathTex(r'id = \int_{Happy}^{Mubarak} \frac{1}{x} \, dx').scale(0.5).shift(UP*1.5)
    id2 = MathTex(r'id =',  r'\ln (x) \Bigr|_{Happy}^{Mubarak}').scale(0.5).shift(UP*1.5)
    self.play(Write(id1))
    self.wait(2)

    #jawab
    jawab = Text('Jawab').scale(0.3).set_color(color='#F38181').shift(UP*2.5)
    self.play(ReplacementTransform(teks, jawab), run_time=0.8)
    self.wait(0.5)

    #ingat 1
    ingat1a = Text('Ingat').scale(0.3).set_color(color='#FCE38A')
    ingat1b = MathTex(r'\int_a^b \frac{1}{x} \, dx =', r'\ln (x) \Bigr|_{a}^{b}').next_to(ingat1a, DOWN, buff=0).scale(0.5).set_color(color='#FCE38A')
    grup_ingat = VGroup(ingat1a, ingat1b)
    kotak1a = SurroundingRectangle(grup_ingat, buff=0.1).set_color(RED)
    kotak1b = SurroundingRectangle(id2[1], buff=0.1).set_color(RED)
    peringatan = VGroup(grup_ingat, kotak1b)
    self.play(Write(grup_ingat), run_time=0.8)
    self.play(Create(kotak1a))
    self.wait()

    self.play(ReplacementTransform(id1, id2))
    self.play(ReplacementTransform(kotak1a, kotak1b), run_time=0.8)
    self.play(FadeOut(peringatan))
    self.wait(0.5)

    #id 3
    id3 = MathTex(r'id =', r'\ln(Mubarak) - \ln(Happy)').scale(0.5).shift(UP*1.5)
    self.play(TransformMatchingShapes(id2, id3))
    self.wait(0.5)

    #ingat 2
    ingat2a = Text('Ingat').scale(0.3).set_color(color='#FCE38A')
    ingat2b = MathTex(r'\ln(a) - \ln(b) =', r'\ln \left( \frac{a}{b} \right)').next_to(ingat2a, DOWN, buff=0).scale(0.5).set_color(color='#FCE38A')
    grup_ingat2 = VGroup(ingat2a, ingat2b)
    kotak2a = SurroundingRectangle(grup_ingat2, buff=0.1).set_color(RED)
    kotak2b = SurroundingRectangle(id3[1], buff=0.1).set_color(RED)
    peringatan2 = VGroup(grup_ingat2, kotak2b)

    self.play(Write(grup_ingat2),  run_time=0.8)
    self.play(Create(kotak2a))
    self.wait()
    self.play(ReplacementTransform(kotak2a, kotak2b), run_time=0.8)
    self.wait()
    self.play(FadeOut(peringatan2))
    self.wait(0.5)

    #id 4, 5
    id4 = MathTex(r'id = \ln \left( \frac{Mubarak}{Happy} \right)').scale(0.5).shift(UP*1.5)
    id5 = MathTex(r'\ln', r'\left(e^{id} \right)', r'=', r' \ln', r'\left( \frac{Mubarak}{Happy} \right)').scale(0.5).shift(UP*1.5)

    #ingat 3
    ingat3a = Text('Ingat').scale(0.3).set_color(color='#FCE38A')
    ingat3b = MathTex(r'\ln(e^a) =', r'a').next_to(ingat3a, DOWN, buff=0).scale(0.5).set_color(color='#FCE38A')
    grup_ingat3 = VGroup(ingat3a, ingat3b)
    kotak3a = SurroundingRectangle(grup_ingat3, buff=0.1).set_color(RED)
    kotak3b = SurroundingRectangle(id5[0:2], buff=0.1).set_color(RED)
    peringatan3 = VGroup(grup_ingat3, kotak3b)

    self.play(TransformMatchingShapes(id3, id4))
    self.wait()
    self.play(TransformMatchingShapes(id4, id5))
    self.play(Write(grup_ingat3),  run_time=0.8)
    self.play(Create(kotak3a))
    self.wait()
    self.play(ReplacementTransform(kotak3a, kotak3b), run_time=0.8)
    self.wait()
    self.play(FadeOut(peringatan3))
    self.wait(0.5)

    #Coret
    coret1 = Cross(id5[0])
    coret2 = Cross(id5[3])
    self.play(Create(coret1), run_time=0.5)
    self.play(Create(coret2), run_time=0.5)
    self.wait(0.5)

    coret3 = VGroup(coret1, coret2, id5)
    #id 6
    id6 = MathTex(r'e^{id} =', r'{Mubarak \over', r'Happy}').scale(0.5).shift(UP*1.5)
    self.play(TransformMatchingShapes(coret3, id6))
    self.wait()

    #id 7
    id7 = MathTex(r'Happy \,\, e^{id}', r'= Mubarak').scale(0.5).shift(UP*1.5)
    self.play(TransformMatchingTex(id6, id7,  key_map={"Happy}": "Happy \,\, e^{id}"}))

    #Happy Eid Mubarak
    mub1 = MathTex(r'Happy \,\, e^{id} \,\, Mubarak').scale(0.5).shift(UP*1.5)
    self.play(TransformMatchingShapes(id7, mub1), FadeOut(jawab))
    mub2 = MathTex(r'Happy \,\, eid \,\, Mubarak').scale(0.5).shift(UP*1.5)
    self.play(TransformMatchingShapes(mub1, mub2))
    self.wait()

    mohon = Text('Mohon Maaf Lahir dan Batin').scale(0.3).shift(UP*1)
    image = ImageMobject("maaf.png").scale(0.5)
    self.play(FadeIn(mohon), FadeIn(image))
    self.wait(2)
```

```python
from manim import *

config.media_width = "70%"

class Project4(Scene):
    def construct(self):
        #Tulis Soal
        soal = MathTex(r"2^{30}+2^{30}+2^{30}+2^{30}=\cdots ?").scale(0.5).shift(UP)
        #Teks
        teks1 = Text('Bagaimana menyelesaikan soal tersebut?', font_size=27, color=RED).scale(0.5).shift(UP*3)
        teks2 = Text('Kotakin aja boy!', color=RED, font_size=30).scale(0.7).shift(UP*3)

        #Animasi Soal
        self.play(Write(soal))
        self.wait(2)
        #Animasi Teks
        self.play(Write(teks1), run_time=1)
        self.play(FadeOut(teks1))
        self.play(GrowFromCenter(teks2))
        self.wait()

        # Misalkan Kotak
        misal = MathTex('Misalkan', r'\rightarrow', color=TEAL).scale(0.5).shift(LEFT*1.2+UP*1.8)
        duapuluh = MathTex(r'2^{30}').scale(0.5).next_to(misal, RIGHT)
        kotak_misalkan = SurroundingRectangle(duapuluh, buff=0.1)
        duapangkattigatiga = MathTex(r"=", r"2^{30}").scale(0.5).next_to(kotak_misalkan, RIGHT, buff=0.1)
        #Animasi Misalkan
        self.play(Write(misal))
        self.play(Write(duapuluh))
        self.play(Write(kotak_misalkan))
        self.play(ReplacementTransform(duapuluh, duapangkattigatiga))
        self.wait()

        #Langkah 1
        langkah1 = MathTex(r'&=',r'2^{30}',r'+',r'2^{30}',r'+',r'2^{30}',r'+',r'2^{30}',       #0-7
                           r'\\ &=',r'2^{30}',r'+',r'2^{30}',r'+',r'2^{30}',r'+',r'2^{30} \\', #8-15
                           r'&= 4 \times \,',r'2^{30} \\',    #16-17
                           r'&= 2^2 \times \,',r'2^{30} \\',  #18-19
                           r'&= 2^2 \times 2^{30} \\',           #20
                           r'&= 2^{32}'                       #21
                           ).scale(0.5).shift(DOWN*0.5)
        #Perkotakan 1
        kotak1a = SurroundingRectangle(langkah1[1], buff=0.05)
        kotak2a = SurroundingRectangle(langkah1[3], buff=0.05)
        kotak3a = SurroundingRectangle(langkah1[5], buff=0.05)
        kotak4a = SurroundingRectangle(langkah1[7], buff=0.05)
        #Animasi Langkah 1
        self.play(ReplacementTransform(soal.copy(), langkah1[0:8]))
        self.wait()
        self.play(Create(kotak1a), run_time=0.5)
        self.play(Create(kotak2a), run_time=0.5)
        self.play(Create(kotak3a), run_time=0.5)
        self.play(Create(kotak4a), run_time=0.5)
        self.wait()
        grup_langkah1 = VGroup(kotak1a, kotak2a, kotak3a, kotak4a)

        #Perkotakan 2
        langkah2 = [langkah1[8], langkah1[10], langkah1[12], langkah1[14]]
        langkah2b = langkah1[9].copy().set_opacity(0)
        langkah2c = langkah1[11].copy().set_opacity(0)
        langkah2d = langkah1[13].copy().set_opacity(0)
        langkah2e = langkah1[15].copy().set_opacity(0)
        kotak1b = SurroundingRectangle(langkah2b, buff=0.05)
        kotak2b = SurroundingRectangle(langkah2c, buff=0.05)
        kotak3b = SurroundingRectangle(langkah2d, buff=0.05)
        kotak4b = SurroundingRectangle(langkah2e, buff=0.05)
        grup_langkah2 = VGroup(langkah2, kotak1b, kotak2b, kotak3b, kotak4b)
        #Animasi langkah 2
        self.play(TransformMatchingShapes(grup_langkah1, grup_langkah2))
        self.wait()

        #Langkah 3
        langkah3 = langkah1[16]
        langkah3b = langkah1[17].copy().set_opacity(0)
        kotak1c = SurroundingRectangle(langkah3b, buff=0.05)
        grup_langkah3 = VGroup(langkah3, kotak1c)
        #Animasi langkah 3
        self.play(Transform(grup_langkah2.copy(), grup_langkah3))
        self.wait()

        #Langkah 4
        langkah4 = langkah1[18]
        langkah4b = langkah1[19].copy().set_opacity(0)
        kotak1d = SurroundingRectangle(langkah4b, buff=0.05)
        grup_langkah4 = VGroup(langkah4, kotak1d)
        #Animasi langkah 4
        self.play(Transform(grup_langkah3.copy(), grup_langkah4))
        self.wait()

        #Langkah 5
        langkah5 = langkah1[20]
        #Animasi langkah 5
        self.play(Transform(grup_langkah4, langkah5))
        self.wait()

        #Langkah 6
        langkah6 = langkah1[21]
        #Animasi langkah 6
        self.play(TransformMatchingShapes(langkah5.copy(), langkah6))
        self.wait(0.5)
        self.play(Circumscribe(langkah6))
        self.wait()

        #Hasil akhir
        jadi = MathTex(r"2^{30}+2^{30}+2^{30}+2^{30}=2^{32}").scale(0.5).shift(UP)
        self.remove(soal)
        self.play(ReplacementTransform(langkah6.copy(), jadi), run_time=1.5)
        self.play(jadi.animate.scale(1.5))

        self.wait(2)
```

```python
from manim import *

config.media_width = "70%"

class Project2(MovingCameraScene):
    def construct(self):
        self.camera.frame.save_state()
        #Judul
        #judul = Title(f'Solusi').scale(0.7).set_color(color='#F38181')
        #self.add(judul)

        #Tulis soal
        hasil_dari = Text('Nilai dari').scale(0.4)
        soal = MathTex(r' \sqrt{\sqrt[3]{\sqrt[3]{3^5 \times 3^6 \times 3^7}}}', r'=', r'\dots?').scale(0.5)
        grup_soal = VGroup(hasil_dari, soal).arrange(RIGHT).set_color(color='#EAFFD0')
        grup_soal.shift(UP*2)
        #Animasi soal
        self.play(Write(grup_soal))
        self.wait(2)

        #Ingat
        ingat = Text('Ingat:').scale(0.3).set_color(color='#FCE38A')
        igt1 = MathTex('\\sqrt[n]{x} = x^{\\frac{1}{n}}').scale(0.5).next_to(ingat, DOWN, buff=0.1).set_color(color='#FCE38A')
        igt2 = MathTex('x^n \\times x^m = x^{n+m}').scale(0.5).next_to(igt1, DOWN, buff=0.1).set_color(color='#FCE38A')
        igt3 = MathTex('(x^n)^m = x^{n \\times m}').scale(0.5).next_to(igt2, DOWN, buff=0.1).set_color(color='#FCE38A')
        grup_ingat = VGroup(ingat, igt1, igt2, igt3)
        grup_ingat.shift(UP*1.2+LEFT*1.2)
        #Animasi ingat
        self.play(Write(ingat))
        self.play(Write(igt1))
        self.wait()
        self.play(Write(igt2))
        self.wait()
        self.play(Write(igt3))
        self.wait()

        #Jawab
        jawab = Text('Jawab:').scale(0.3).next_to(igt3, DOWN).set_color(color='#EAFFD0')
        self.play(DrawBorderThenFill(jawab))


        #Langkah penyelesaian
        l = MathTex(r' \sqrt{ \sqrt[3]{\sqrt[3]{3^5 \times 3^6 \times 3^7}}}').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')
        l1 = MathTex(r'\left( \sqrt[3]{\sqrt[3]{3^5 \times 3^6 \times 3^7}} \right)^\frac{1}{2}').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')
        l2 = MathTex(r'\left( \left(\sqrt[3]{3^5 \times 3^6 \times 3^7} \right)^\frac{1}{3} \right)^\frac{1}{2}').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')
        l3 = MathTex(r'\left(\left(\left(3^5 \times 3^6 \times 3^7 \right)^\frac{1}{3} \right)^\frac{1}{3} \right)^\frac{1}{2}').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')
        l4 = MathTex(r'\left(\left(\left(3^{5+6+7} \right)^\frac{1}{3} \right)^\frac{1}{3} \right)^\frac{1}{2}').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')
        l5 = MathTex(r'\left(\left(\left(3^{18} \right)^\frac{1}{3} \right)^\frac{1}{3} \right)^\frac{1}{2}').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')
        l6 = MathTex(r'\left(\left(3^{18} \right)^\frac{1}{3} \right)^\frac{1}{6}\\').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')
        l7 = MathTex(r'\left(3^{18} \right)^\frac{1}{18}\\').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')
        l8 = MathTex(r'3^{\frac{18}{18}}').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')
        l9 = MathTex(r'3^1').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')
        l10 = MathTex(r'3').scale(0.5).shift(DOWN).set_color(color='#EAFFD0')

        self.play(ReplacementTransform(soal[0].copy(), l), run_time=1.5)
        self.wait()
        self.play(self.camera.frame.animate.scale(0.8).move_to(l))
        self.remove(l)
        self.play(Transform(l, l1))
        self.wait()
        self.remove(l, l1)
        self.play(Transform(l1, l2))
        self.wait()
        self.remove(l, l1, l2)
        self.play(Transform(l2, l3))
        self.wait()
        self.remove(l, l1, l2, l3)
        self.play(Transform(l3, l4))
        self.wait()
        self.remove(l, l1, l2, l3, l4)
        self.play(Transform(l4, l5))
        self.wait()
        self.remove(l, l1, l2, l3, l4, l5)
        self.play(Transform(l5, l6))
        self.wait()
        self.remove(l, l1, l2, l3, l4, l5, l6)
        self.play(Transform(l6, l7))
        self.wait()
        self.remove(l, l1, l2, l3, l4, l5, l6, l7)
        self.play(Transform(l7, l8))
        self.wait()
        self.remove(l, l1, l2, l3, l4, l5, l6, l7, l8)
        self.play(Transform(l8, l9))
        self.remove(l, l1, l2, l3, l4, l5, l6, l7, l8, l9)
        self.play(Transform(l9, l10))
        self.play(Restore(self.camera.frame))
        self.play(Indicate(l10))
        self.wait()
        self.remove(l, l1, l2, l3, l4, l5, l6, l7, l8, l9, l10)

        #Hasil akhir
        self.add(grup_soal)
        jadi = MathTex(r' \sqrt{\sqrt[3]{\sqrt[3]{3^5 \times 3^6 \times 3^7}}}', r'=3').scale(0.5).next_to(hasil_dari, RIGHT)
        self.remove(soal)
        self.play(ReplacementTransform(Group(l10, soal), jadi))
        self.wait()
        self.remove(jawab, soal, ingat, igt1, igt2, igt3, jadi, hasil_dari)
        self.play(jadi.animate.shift(DOWN+LEFT*0.5).scale(1.5))

        self.wait()
```

## Video 8 | Heart Equation

```python
from manim import *
import numpy as np

class HeartEquation(Scene):
    def construct(self):
      #Judul
      tex = Tex(r"$\mathbb{H}eart \, \mathbb{E}quation$").set_color_by_gradient(RED, GOLD).scale(0.7)
      ul = Underline(tex).set_color(GREY)
      judul = VGroup(tex, ul)
      judul.to_edge(UP)

      #Sumbu Koordinat
      axes = Axes(x_range=[-2, 2, 1],
                  y_range=[-1.5, 2.5, 1],
                  x_length=5,
                  y_length=5,
                  tips=True)
      axes.scale(0.7).next_to(judul, DOWN)

      #Fungsi
      a = ValueTracker(0)
      def fungsi():
          return axes.plot(lambda x: np.cbrt(x)**2 + 0.9 * np.sin(a.get_value()*x) * np.sqrt(3 - x**2),
                           x_range=[-1.73, 1.73], color=RED)

      grafik = always_redraw(fungsi)

      #Persamaan
      y = MathTex(r'y=x^{\frac{2}{3}} + 0.9 \sin(kx) \sqrt{3-x^2}').scale(0.6).next_to(axes, DOWN)
      # k = Variable(0, 'k', num_decimal_places=1).scale(0.7).next_to(y, DOWN)
      # k.label.set_color(RED)
      # k.value.set_color(WHITE)

      def update_label(mobj):
          k_val = a.get_value()
          new_tex = MathTex(
              rf"y = x^{{\frac{{2}}{{3}}}} + 0.9 \sin({k_val:.1f}x) \sqrt{{3-x^2}}").scale(0.6).next_to(axes, DOWN)
          mobj.become(new_tex)

      y.add_updater(update_label)

      #Animasi
      self.add(judul)
      self.play(Write(y), GrowFromCenter(axes))
      self.play(Create(grafik))

      self.play(a.animate.set_value(21), run_time=7)
      self.wait()

```

```python
import numpy as np
import matplotlib.pyplot as plt

# Definisikan fungsi akar pangkat tiga yang benar untuk nilai negatif
def cube_root(x):
    return np.sign(x) * (np.abs(x) ** (1/3))

# Domain x, pastikan batas -√3 <= x <= √3 agar sqrt(3 - x^2) valid
x = np.linspace(-np.sqrt(3), np.sqrt(3), 1000)

# Hitung fungsi
y = cube_root(x)**2 + 0.9 * np.sin(30 * x) * np.sqrt(3 - x**2)

# Plot
plt.plot(x, y)
plt.xlabel('x')

plt.title('Grafik Hati')
plt.grid(True)
plt.show()
```

```python
from manim import *

class Parabola(Scene):
    def construct(self):
        axes = Axes(
            x_range=[-3, 3, 1],
            y_range=[-3, 3, 1],
            x_length=10,
            y_length=6,
        ).to_edge(DOWN)
        self.play(Create(axes))

        # ValueTracker a
        a = ValueTracker(2)

        # Label awal: isi a nanti akan diganti dinamis
        persamaan = MathTex("f(x) = -2.00x^2").to_edge(UP)

        def update_label(mobj):
            new_a = a.get_value()
            new_tex = MathTex(f"f(x) = {new_a:.1f}x^2")
            mobj.become(new_tex.to_edge(UP))

        persamaan.add_updater(update_label)
        self.play(Write(persamaan))

        # Plot fungsi dinamis
        graph = always_redraw(lambda:
            axes.plot(lambda x: a.get_value() * x**2, color=YELLOW)
        )
        self.play(Create(graph))

        # Animasi: ubah a
        self.play(a.animate.set_value(-2), run_time=4)
        self.wait()
        self.play(a.animate.set_value(2), run_time=4)
        self.wait(2)
```

```python
from manim import *

config.media_width = "70%"

class TeoremaPythagoras(Scene):
    def construct(self):
      #Judul
      judul = Text('Teorema Pythagoras', font_size=48)

      self.play(Write(judul))
      self.wait(0.5)
      self.play(FadeOut(judul))

      A = 1.5 * RIGHT + 0.5 * DOWN
      B = 1.5 * RIGHT + 0.5 * UP
      C = 1.5 * LEFT + 0.5 * DOWN
      D = A + LEFT * 0.15 + UP * 0.15
      E = A + LEFT * 0.15
      F = A + UP * 0.15

      segitiga = Polygon(A, B, C, color=BLUE)
      segitiga.set_fill(BLUE, opacity=0.4)

      sudut = Polygon(A, E, D, F, color=BLUE)

      self.play(Create(segitiga), Create(sudut))
      self.wait(0.5)

      a = MathTex('a').next_to(segitiga[0], DOWN)
      b = MathTex('b').next_to(segitiga[0], RIGHT)
      c = MathTex('c').next_to(segitiga.get_center(), UP)

      self.play(Write(a), Write(b), Write(c))
      self.wait(0.5)

      segitiga3 = segitiga.copy()
      segitiga9 = segitiga.copy()
      segitiga12 = segitiga.copy()

      segitiga3.set_fill(RED, opacity=0.4).set_color(RED)
      segitiga9.set_fill(GREEN, opacity=0.4).set_color(GREEN)
      segitiga12.set_fill(YELLOW, opacity=0.4).set_color(YELLOW)

      self.play(
          FadeOut(a), FadeOut(b), FadeOut(c), FadeOut(sudut),
          segitiga3.animate.move_to(UP * .5 + RIGHT * 1.5),
          segitiga.animate.move_to(DOWN * 1.5 + RIGHT * .5),
          segitiga9.animate.move_to(LEFT * 1.5 + DOWN * .5),
          segitiga12.animate.move_to(UP * 1.5 + LEFT * .5)
      )

      a0 = MathTex('a').move_to(segitiga9.get_center() + LEFT *1.5/2)
      b0 = MathTex('b').move_to(segitiga12.get_center() + LEFT *3.5/2)
      a1 = MathTex('a').move_to(segitiga12.get_center() + UP *1.5/2)
      b1 = MathTex('b').move_to(segitiga3.get_center() + UP *3.5/2)
      c1 = MathTex('c').move_to(segitiga.get_center() + UP *0.5/2)
      c2 = MathTex('c').move_to(segitiga9.get_center() + RIGHT *0.5/2)
      c3 = MathTex('c').move_to(segitiga12.get_center() + DOWN *0.5/2)
      c4 = MathTex('c').move_to(segitiga3.get_center() + LEFT *0.5/2)

      kotak = Square(np.sqrt(10), color=WHITE, fill_opacity=0.4).move_to(ORIGIN).rotate(np.arctan(1/3))
      kotak2 = Square(4, color=RED, fill_opacity=0.4).move_to(DOWN*1)

      self.play(
          segitiga3.animate.rotate(PI/2),
          segitiga9.animate.rotate(-PI/2),
          segitiga12.animate.rotate(PI),
          Create(kotak)
      )
      self.play(
          Write(b0), Write(a0), Write(a1), Write(b1),
          Write(c1), Write(c2), Write(c3), Write(c4)
      )
      self.wait(0.5)

      ab1 = MathTex('a + b').move_to(segitiga9.get_center() + UP * 1/2 + LEFT * 2/2)
      ab2 = MathTex('a + b').move_to(segitiga12.get_center() + UP * 1.5/2 + RIGHT * 1/2)
      ckuadrat = MathTex('c^2').move_to(ORIGIN)

      a00 = VGroup(a0, b0)
      a11= VGroup(a1, b1)
      c = VGroup(c1, c2, c3, c4)
      triangles = VGroup(segitiga, segitiga3, segitiga9, segitiga12, kotak)

      self.play(
          Transform(a00, ab1), Transform(a11, ab2), Transform(c, ckuadrat),
          Indicate(kotak, color=WHITE, scale_factor=1.5)
      )

      self.play(
          c.animate.move_to(UP*6/2),
          triangles.animate.move_to(DOWN*2/2),
          a00.animate.move_to(LEFT*5/2 + DOWN*2/2),
          a11.animate.move_to(UP*2.5/2)
      )

      self.wait(0.5)

      after = MathTex(r'c^2 + \frac{4ab}{2}').move_to(UP*6/2)
      after2 = MathTex(r'c^2 + 2ab = (a+b)^2').move_to(UP*6/2)
      after3 = MathTex(r'c^2 + 2ab = a^2 + 2ab + b^2')
      after4 = MathTex(r'c^2 = b^2 + a^2')

      frame = SurroundingRectangle(after4, color=PURE_RED, buff=0.3)

      a0011 = VGroup(a00, a11, c)

      self.play(
          Indicate(segitiga, color=BLUE, scale_factor=1.2),
          Indicate(segitiga3, color=RED, scale_factor=1.2),
          Indicate(segitiga9, color=GREEN, scale_factor=1.2),
          Indicate(segitiga12, color=YELLOW, scale_factor=1.2),
          Transform(a0011, after2)
      )
      self.wait(0.5)

      self.play(DrawBorderThenFill(kotak2))
      self.play(FadeOut(triangles))
      self.play(
          Indicate(kotak2, color=PURE_RED, scale_factor=1.2),
          Transform(a0011, after2)
      )
      self.wait()

      self.play(Transform(a0011, after3), FadeOut(kotak2))
      self.wait()

      self.play(Transform(a0011, after4), Create(frame))
      self.wait(3)
```

```python
from manim import *

class VisualizeGradient1D(Scene):
    def construct(self):
        # 1. Pengaturan Dasar
        # --------------------
        # Membuat sumbu koordinat
        axes = Axes(
            x_range=[-5, 5, 1],  # Rentang sumbu x [awal, akhir, step]
            y_range=[-1, 7, 1],  # Rentang sumbu y
            x_length=10,         # Panjang sumbu x di layar
            y_length=6,          # Panjang sumbu y di layar
            axis_config={"include_numbers": True, "font_size": 24},
        )
        x_label = axes.get_x_axis_label("x")
        y_label = axes.get_y_axis_label("f(x)")
        self.add(axes, x_label, y_label)

        # 2. Definisi Fungsi dan Turunannya
        # ----------------------------------
        # Fungsi f(x) = x^2 / 4
        def func(x):
            return 0.25 * x**2

        # Turunan f'(x) = x / 2 (ini adalah gradien)
        def deriv_func(x):
            return 0.5 * x

        # Rentang untuk plot grafik fungsi
        x_plot_range = [-4, 4]

        # 3. Membuat Grafik Fungsi
        # -------------------------
        graph = axes.plot(func, x_range=x_plot_range, color=BLUE)
        graph_label = axes.get_graph_label(graph, label=MathTex("f(x) = \\frac{x^2}{4}"), x_val=x_plot_range[1], direction=UR)
        self.add(graph, graph_label)

        # 4. ValueTracker untuk Mengontrol Posisi Titik x
        # ------------------------------------------------
        # ValueTracker memungkinkan kita mengubah nilai x secara dinamis
        x_tracker = ValueTracker(-3)  # Nilai x awal

        # 5. Membuat Titik pada Kurva dan Garis Singgung (Tangen)
        # -------------------------------------------------------
        # Titik pada kurva yang akan bergerak
        # axes.c2p mengubah koordinat grafik menjadi koordinat layar Manim
        moving_dot = Dot(color=RED)
        moving_dot.add_updater(
            lambda mob: mob.move_to(axes.c2p(x_tracker.get_value(), func(x_tracker.get_value())))
        )

        # Fungsi untuk membuat objek garis singgung
        # always_redraw akan menggambar ulang objek ini setiap frame
        def get_tangent_line():
            x_val = x_tracker.get_value()
            y_val = func(x_val)
            slope = deriv_func(x_val)

            # Buat segmen garis untuk tangen
            # y - y1 = m(x - x1) => y = m(x - x1) + y1
            # Kita butuh dua titik untuk objek Line.
            # Pilih titik x +/- delta dari x_val
            delta_x_tangent = 1.5  # Setengah panjang visual dari garis singgung

            x1 = x_val - delta_x_tangent
            y1_line = slope * (x1 - x_val) + y_val

            x2 = x_val + delta_x_tangent
            y2_line = slope * (x2 - x_val) + y_val

            return Line(axes.c2p(x1, y1_line), axes.c2p(x2, y2_line), color=YELLOW, stroke_width=3.5)

        tangent_line = always_redraw(get_tangent_line)

        # 6. Menampilkan Nilai Gradien (Slope)
        # -------------------------------------
        slope_text_intro = MathTex("f'(x) = ", font_size=36).to_corner(UR, buff=0.5)
        slope_value_text = DecimalNumber(
            deriv_func(x_tracker.get_value()),
            num_decimal_places=2,
            font_size=36
        )
        # Updater untuk nilai gradien
        slope_value_text.add_updater(
            lambda mob: mob.set_value(deriv_func(x_tracker.get_value())).next_to(slope_text_intro, RIGHT, buff=0.1)
        )

        # 7. Visualisasi dx dan dy (Segitiga Kecil)
        # ----------------------------------------
        dx_visual_length = 0.75 # Panjang visual untuk dx pada grafik

        # Garis untuk dx
        dx_line = always_redraw(lambda: Line(
            axes.c2p(x_tracker.get_value(), func(x_tracker.get_value())), # Titik P
            axes.c2p(x_tracker.get_value() + dx_visual_length, func(x_tracker.get_value())), # Titik (x+dx, y)
            color=GREEN, stroke_width=3
        ))

        # Garis untuk dy
        dy_line = always_redraw(lambda: Line(
            axes.c2p(x_tracker.get_value() + dx_visual_length, func(x_tracker.get_value())), # Titik (x+dx, y)
            axes.c2p(x_tracker.get_value() + dx_visual_length, func(x_tracker.get_value()) + deriv_func(x_tracker.get_value()) * dx_visual_length), # Titik (x+dx, y+dy)
            color=ORANGE, stroke_width=3
        ))

        # Label untuk dx dan dy
        dx_label = MathTex("dx", font_size=24, color=GREEN)
        dx_label.add_updater(lambda m: m.next_to(dx_line, DOWN, buff=0.1))

        dy_label = MathTex("dy", font_size=24, color=ORANGE)
        dy_label.add_updater(lambda m: m.next_to(dy_line, RIGHT if deriv_func(x_tracker.get_value()) >= 0 else LEFT, buff=0.1))

        # 8. Menambahkan Objek ke Scene dan Animasi
        # -----------------------------------------
        self.add(moving_dot, tangent_line, slope_text_intro, slope_value_text)
        self.add(dx_line, dy_line, dx_label, dy_label)

        # Animasi: gerakkan titik x dari -3 ke 3
        self.play(
            x_tracker.animate.set_value(3),
            run_time=7,      # Durasi animasi dalam detik
            rate_func=linear # Kecepatan animasi (bisa juga smooth, there_and_back, dll.)
        )
        self.wait(2) # Tunggu 2 detik di akhir
```

# Heart Curve

```python
from manim import *
import numpy as np

class HeartCurve(Scene):
    def construct(self):
      #Judul
      teks = Tex(r"$\mathbb{H}eart \, \mathbb{C}urve$").set_color_by_gradient(PINK, PURE_RED).scale(0.7)

      #Sumbu Koordinat
      axes = Axes(x_range=[-2, 2, 1],
                  y_range=[-1.5, 2.5, 1],
                  x_length=5,
                  y_length=5,
                  tips=True)
      axes.scale(0.7).shift(UP*0.7)

      a = ValueTracker(0)
      #Grafik
      grafik = always_redraw(
           lambda: axes.plot(lambda x: np.cbrt(x)**2 + 0.9 * np.sin(a.get_value()*x) * np.sqrt(3 - x**2),
                           x_range=[-1.73, 1.73], color=PURE_RED)
           )

      #Persamaan
      equation = always_redraw(
            lambda: MathTex(
                rf"y = x^{{\frac{{2}}{{3}}}} + 0.9 \sin({a.get_value():.1f}x) \sqrt{{3-x^2}}"
            ).scale(0.6).next_to(axes, DOWN).set_color_by_gradient(PURE_RED, PINK)
        )

      #Animasi
      self.play(Write(teks))
      self.play(teks.animate.shift(UP*3), GrowFromCenter(axes), Create(grafik))
      self.play(DrawBorderThenFill(equation))

      self.play(a.animate.set_value(20), run_time=3)
      self.wait(0.1)
      self.play(a.animate.set_value(0), run_time=3)
      self.wait(0.1)
      self.play(a.animate.set_value(20), run_time=3)
      self.wait()

      box = SurroundingRectangle(equation, corner_radius=0.2).set_color(PURE_RED)
      self.play(Create(box))
      self.wait(0.5)
      grup = VGroup(teks, axes, grafik, equation, box)

      #Watermark
      rect = RoundedRectangle(corner_radius=0.3, height=1.5, width=1.5, color=PINK)
      obj = Circle(radius=0.4, color=PINK).next_to(rect, DOWN, buff=-1.15)
      dot = Dot(color=PINK).scale(0.6).next_to(obj, buff=0.15).shift(UP*0.5)
      ig = VGroup(rect, obj, dot).scale(0.4)
      my_id = Text("@na_dzirrr", color=PINK, font_size=20).next_to(ig, buff=0.1)
      wm = VGroup(ig, my_id).move_to(ORIGIN)

      quotes = Text("Mathematics and love are complicated,\nbut there is always beauty in them"
                    , font="Open Sans", line_spacing=1, slant=ITALIC, color=RED, font_size=17).next_to(wm, DOWN, buff=0.3)

      self.play(Transform(grup, wm), Write(quotes))
      self.add(wm)
      self.remove(equation)
      self.remove(grafik)

      self.wait(2)
```

```python
from manim import *
config.media_width = "70%"

class Scene2(Scene):
  def construct(self):
    #Frame
    titik1 = [-2.2, 4, 0]
    titik2 = [2.2, 4, 0]
    titik3 = [2.2, -4, 0]
    titik4 = [-2.2, -4, 0]
    frame_asli = Polygon(titik1, titik2, titik3, titik4, color=WHITE, stroke_width=2)
    self.add(frame_asli)
    # --- 0. Setup Awal ---
    a = 2
    b = 1

    #Watermark
    rect = RoundedRectangle(corner_radius=0.3, height=1.5, width=1.5, color=PINK)
    obj = Circle(radius=0.4, color=PINK).next_to(rect, DOWN, buff=-1.15)
    dot = Dot(color=PINK).scale(0.6).next_to(obj, buff=0.15).shift(UP*0.4)
    ig = VGroup(rect, obj, dot).scale(0.3).move_to(ORIGIN)
    my_id = Text("@na_dzirrr", font_size=27).next_to(ig, buff=0.1)
    wm = VGroup(ig, my_id).scale(0.4).move_to(ORIGIN)

    # PENDAHULUAN
    rumus = MathTex('(a+b)^2 =', 'a^2', '+', '2ab', '+', 'b^2', font_size=35)

    self.play(Write(rumus),
              FadeIn(my_id), Create(ig),
              wm.animate.scale(0.4).move_to(UP*3.5))
    self.wait()
    self.play(rumus.animate.shift(UP*2))
    self.wait()
    # ---- SCENE 2 ----
    rumus = MathTex('(a+b)^2', '= a^2', '+', '2ab', '+', 'b^2', font_size=35)
    self.add(rumus.shift(UP*2), wm)

    # --- 1. Membuat garis a dan b ---
    garis_a = Line(ORIGIN, RIGHT * a, color=BLUE, stroke_width=2)
    garis_b = Line(RIGHT * a, RIGHT * (a + b), color=RED, stroke_width=2)

    garis_a.move_to(UL*0.3)
    garis_b.next_to(garis_a, RIGHT)
    # Membuat Brace pertama
    brace_a = Brace(garis_a, direction=UP, buff=0.1, stroke_width=0)
    teks_brace_a = brace_a.get_tex('a', buff=0.1)
    brace_b = Brace(garis_b, direction=UP, buff=0.1, stroke_width=0)
    teks_brace_b = brace_b.get_tex('b', buff=0.1)
    grup_brace1 = VGroup(teks_brace_a, teks_brace_b)

    garis_b2 = garis_b.copy().next_to(garis_a, RIGHT, buff=0)
    # Brace gabungan
    brace_ab = BraceBetweenPoints(garis_a.get_start(), garis_b2.get_end(), buff=0.1, direction=UP, stroke_width=0)
    teks_brace_ab = brace_a.get_tex('a+b', buff=0).shift(RIGHT*.45).scale(0.7)

    grup_brace_ab = VGroup(brace_ab, teks_brace_ab)

    # Membuat Sisi kiri
    garis_ab = VGroup(garis_a, garis_b2)
    sisi_kiri = garis_ab.copy().rotate(-PI/2).move_to(garis_a.get_start(), UP)
    brace_sisi_kiri = Brace(sisi_kiri, direction=LEFT, buff=0.1, stroke_width=0)
    teks_brace_sisi_kiri = brace_sisi_kiri.get_tex('a+b', buff=0.1).rotate(PI/2).shift(RIGHT*0.4).scale(0.7)
    grup_brace_kiri = VGroup(brace_sisi_kiri, teks_brace_sisi_kiri)

    # Kumpulan Persegi
    persegi = Square(side_length=a+b, stroke_width=1).move_to(garis_a.get_start(), UL).set_fill(color=DARK_BLUE, opacity=0.5)
    persegi1 = Square(side_length=a, color=BLUE, fill_opacity=0.5, stroke_width=2).move_to(garis_a.get_center(), UP)
    persegi2 = Rectangle(height=a, width=b, color=RED, fill_opacity=0.5, stroke_width=2).move_to(persegi1.get_vertices()[0], UL)
    persegi3 = Rectangle(height=b, width=a, color=RED, fill_opacity=0.5, stroke_width=2).move_to(persegi1.get_vertices()[2], UL)
    persegi4 = Square(side_length=b, color=PURPLE, fill_opacity=0.5, stroke_width=2).move_to(persegi1.get_vertices()[3], UL)

    # Teks a + b
    ab_kuadrat = MathTex('(a+b)^2', font_size=35).move_to(persegi.get_center())

    # Persamaan (a+b)^2
    pers = MathTex('(a+b)^2=', 'a^2', '+ ab + ab', '+ b^2', font_size=33).next_to(rumus, DOWN)

    self.play(ReplacementTransform(rumus[0].copy(), grup_brace1),
                Create(brace_a), Create(brace_b), Create(garis_a), Create(garis_b))
    self.wait()

    self.play(ReplacementTransform(garis_b, garis_b2),
              ReplacementTransform(Group(brace_a, brace_b), brace_ab),
              ReplacementTransform(grup_brace1, teks_brace_ab))
    self.wait()

    self.play(FadeIn(persegi),
              ReplacementTransform(grup_brace_ab.copy(), grup_brace_kiri))
    self.wait()

    self.play(ReplacementTransform(Group(grup_brace_ab, grup_brace_kiri), ab_kuadrat))
    self.wait()

    self.play(ReplacementTransform(ab_kuadrat, pers[0]))

    # Label Persegi a^2
    brace_a = Brace(garis_a, direction=UP, buff=0.1, stroke_width=0)
    teks_brace_a = brace_a.get_tex('a', buff=0.1)
    brace_a_kiri = Brace(sisi_kiri[0], direction=LEFT, buff=0.1, stroke_width=0)
    teks_brace_a_kiri = brace_a_kiri.get_tex('a', buff=0.1)

    # Label Persegi a dan b kiri
    brace_b = Brace(garis_b, direction=UP, buff=0.1, stroke_width=0)
    teks_brace_b = brace_b.get_tex('b', buff=0.1)
    brace_b_kiri = Brace(sisi_kiri[1], direction=LEFT, buff=0.1, stroke_width=0)
    teks_brace_b_kiri = brace_b_kiri.get_tex('b', buff=0.1)

    # Teks a^2, b^2, ab
    aa = MathTex('a^2', font_size=35).move_to(persegi1.get_center())
    ab1 = MathTex('ab', font_size=35).move_to(persegi2.get_center())
    ab2 = MathTex('ab', font_size=35).move_to(persegi3.get_center())
    bb = MathTex('b^2', font_size=35).move_to(persegi4.get_center())

    persegi.set_z_index(-1) #Mengubah persegi besar supaya di belakang
    bagian_a = VGroup(brace_a, teks_brace_a, brace_a_kiri, teks_brace_a_kiri)
    self.play(FadeIn(bagian_a),
              Create(persegi1))
    self.wait()

    self.play(ReplacementTransform(Group(teks_brace_a, teks_brace_a_kiri), aa),
              FadeOut(Group(brace_a, brace_a_kiri)))
    self.wait()

    self.play(ReplacementTransform(aa.copy(), pers[1]))
    self.wait()

    # Label Persegi a^2
    brace_a = Brace(garis_a, direction=UP, buff=0.1, stroke_width=0)
    teks_brace_a = brace_a.get_tex('a', buff=0.1)
    brace_a_kiri = Brace(sisi_kiri[0], direction=LEFT, buff=0.1, stroke_width=0)
    teks_brace_a_kiri = brace_a_kiri.get_tex('a', buff=0.1)
    self.play(Create(persegi2), Create(persegi3),
              Create(brace_b), Create(teks_brace_b),
              Create(brace_b_kiri), Create(teks_brace_b_kiri),
              Group(brace_a, teks_brace_a).animate.shift(DOWN*a),
              Group(brace_a_kiri, teks_brace_a_kiri).animate.shift(RIGHT*a))
    self.wait()

    self.play(ReplacementTransform(Group(teks_brace_b, teks_brace_a), ab1),
              ReplacementTransform(Group(teks_brace_b_kiri, teks_brace_a_kiri), ab2),
              FadeOut(Group(brace_b, brace_b_kiri, brace_a, brace_a_kiri)))
    self.wait()

    self.play(ReplacementTransform(Group(ab1.copy(), ab2.copy()), pers[2]))
    self.wait()

    # Label Persegi a^2
    brace_a = Brace(garis_a, direction=UP, buff=0.1, stroke_width=0)
    teks_brace_a = brace_a.get_tex('a', buff=0.1)
    brace_a_kiri = Brace(sisi_kiri[0], direction=LEFT, buff=0.1, stroke_width=0)
    teks_brace_a_kiri = brace_a_kiri.get_tex('a', buff=0.1)
    # Label Persegi a dan b kiri
    brace_b = Brace(garis_b, direction=UP, buff=0.1, stroke_width=0)
    teks_brace_b = brace_b.get_tex('b', buff=0.1)
    brace_b_kiri = Brace(sisi_kiri[1], direction=LEFT, buff=0.1, stroke_width=0)
    teks_brace_b_kiri = brace_b_kiri.get_tex('b', buff=0.1)

    self.play(Create(persegi4),
              Create(brace_b), Create(teks_brace_b),
              Create(brace_b_kiri), Create(teks_brace_b_kiri),
              Group(brace_b, teks_brace_b).animate.shift(DOWN*a),
              Group(brace_b_kiri, teks_brace_b_kiri).animate.shift(RIGHT*a))
    self.wait()

    self.play(ReplacementTransform(Group(teks_brace_b, teks_brace_b_kiri), bb),
              FadeOut(Group(brace_b, brace_b_kiri)))
    self.wait()

    self.play(ReplacementTransform(bb.copy(), pers[3]))

    pers2 = MathTex('(a+b)^2=', 'a^2', '+', '2ab', '+', 'b^2', font_size=33).next_to(rumus, DOWN)
    self.play(TransformMatchingTex(pers, pers2, key_map={'ab + ab':'2ab'}))

    self.play(ReplacementTransform(pers2, rumus))
    self.play(Circumscribe(rumus))
    self.wait()
```

```python
from manim import *
class SoalCPNS(Scene):
  def construct(self):
    # --- Header ---
    header = Text("Latihan Soal TIU CPNS", weight=BOLD, font_size=40).set_color_by_gradient(RED, GOLD)
    header.to_edge(UP)
    garis = Line(LEFT*3.5, RIGHT*3.5, color=WHITE).next_to(header, DOWN, buff=0.2)
    self.play(Write(header), Create(garis), run_time=.5)

    # --- Soal ---
    soal = Text('7 adalah berapa persen dari 20?', font_size=30).shift(UP*5)
    a = Text('A. 20%', font_size=30).shift(UP*4.2 + LEFT*2.7)
    b = Text('B. 15%', font_size=30).next_to(a, DOWN).align_to(a, LEFT)
    c = Text('C. 35%', font_size=30).next_to(b, DOWN).align_to(a, LEFT)
    d = Text('D. 3,5%', font_size=30).next_to(c, DOWN).align_to(a, LEFT)
    e = Text('E. 0,35%', font_size=30).next_to(d, DOWN).align_to(a, LEFT)
    grup_soal = VGroup(soal, a,b,c,d,e)
    self.play(FadeIn(grup_soal, shift=UP))
    self.wait()

    # --- Animasi silahkan kerjakan ---
    kerjakan = Text('Silahkan pause dan kerjakan...', font_size=30).set_color_by_gradient(RED, GOLD)
    bahas = Text('Mari kita bahas', font_size=30).set_color_by_gradient(RED, GOLD)
    self.play(Write(kerjakan))
    self.wait()
    for i in range(3, 0, -1):
      angka = Text(str(i), font_size=90).shift(DOWN*2).set_color_by_gradient(RED, GOLD)
      self.play(GrowFromCenter(angka), run_time=0.7)
      self.play(FadeOut(angka), run_time=0.2)
    self.play(ReplacementTransform(kerjakan, bahas))
    self.play(FadeOut(bahas, scale=2))
    self.wait()

    # --- Rumus ----
    rumus = Text('Rumus Persentase', font_size=30, color=RED).next_to(e, DOWN, buff=0.7).align_to(a, LEFT)
    persen = MathTex(r'\text{persentase}=\frac{\text{bagian}}{\text{keseluruhan}} \times 100\%', color=YELLOW, font_size=30).next_to(rumus, DOWN, buff=0.5).align_to(rumus, LEFT)
    kotak = SurroundingRectangle(persen, buff=0.2, corner_radius=0.2).set_color_by_gradient(RED, GOLD)
    self.play(FadeIn(Group(rumus, persen, kotak), shift=UP))
    self.wait()

    # --- Pembahasan ---
    self.play(Indicate(soal[0], scale_factor=2, color=BLUE), Indicate(soal[23:25], scale_factor=2, color=RED), run_time=1.5)

    langkah1 = MathTex(r'\frac{7}{20} \times 100\%').next_to(kotak, DOWN, buff=1).scale(1.5)
    langkah2 = MathTex(r'7 \times 5\%').next_to(kotak, DOWN, buff=1).scale(1.5)
    langkah1[0][0].set_color(BLUE) #warnain 7 jadi biru
    langkah1[0][2:4].set_color(RED) #warnain 20 jadi merah
    coret1 = Line(start=langkah1[0][5:8].get_corner(DL), end=langkah1[0][5:8].get_corner(UR), color=RED)
    coret2 = Line(start=langkah1[0][2:4].get_corner(DL), end=langkah1[0][2:4].get_corner(UR), color=RED)
    coret = VGroup(coret1, coret2)
    lima = MathTex('5').next_to(langkah1[0][5:], UP).scale(1.5)
    darisoal = VGroup(soal[0], soal[23:25])

    self.play(TransformFromCopy(darisoal, langkah1[0][:4]))
    self.play(FadeIn(langkah1[0][4:], shift=LEFT))
    self.play(Create(coret1), Create(coret2))
    self.play(TransformFromCopy(langkah1[0][5:8], lima))
    transformasikan = VGroup(coret, lima, langkah1)
    self.play(TransformMatchingShapes(transformasikan, langkah2))

    langkah3 = MathTex(r'35\%').next_to(kotak, DOWN, buff=1).scale(1.5)
    self.play(TransformMatchingShapes(langkah2, langkah3))
    self.play(Circumscribe(langkah3))
    self.play(c.animate.scale(2).set_color(YELLOW))

    self.wait(2)

    # --- Hilangkan yang ada di layar ---
    hilangkan = VGroup(soal, a,b,c,d,e, rumus, persen, kotak, langkah3)
    self.play(FadeOut(hilangkan, shift=DOWN))

    # --- Soal untuk kalian ---
    soal_foru = Text('Soal untuk kalian', font_size=30).shift(UP*5.8 + LEFT*1.6).set_color_by_gradient(RED, GOLD)
    soal = Text('24 adalah berapa persen dari 80?', font_size=30).shift(UP*5).align_to(soal_foru, LEFT)
    a = Text('A. 30%', font_size=30).shift(UP*4.2 + LEFT*2.7)
    b = Text('B. 15%', font_size=30).next_to(a, DOWN).align_to(a, LEFT)
    c = Text('C. 20%', font_size=30).next_to(b, DOWN).align_to(a, LEFT)
    d = Text('D. 24%', font_size=30).next_to(c, DOWN).align_to(a, LEFT)
    e = Text('E. 50%', font_size=30).next_to(d, DOWN).align_to(a, LEFT)
    soal_untukmu = VGroup(soal_foru, soal, a,b,c,d,e)
    self.play(Write(soal_untukmu))
    self.wait()
```

```python
from manim import *

class SolusiPersentase(Scene):
    def construct(self):
        # --- BAGIAN 1: MENAMPILKAN SOAL ---

        # Header
        header = Text("TRIK TIU CPNS", font_size=60, weight=BOLD, color=YELLOW).to_edge(UP, buff=1)

        # Teks Soal (Kita susun agar mudah dibaca di layar HP)
        soal_teks = VGroup(
            Text("4/5 dari 15", font_size=50),
            Text("akan sama dengan", font_size=40),
            Text("30% dari ...", font_size=50, color=BLUE)
        ).arrange(DOWN, buff=0.3).next_to(header, DOWN, buff=1)

        self.play(Write(header), FadeIn(soal_teks, shift=UP))
        self.wait(1)

        # --- BAGIAN 2: MENERJEMAHKAN KE MATEMATIKA ---

        # Kita buat persamaan matematika yang posisinya pas di bawah teks soal
        # Format: [0]=4/5, [1]=x, [2]=15, [3]=, [4]=30%, [5]=x, [6]=...
        persamaan_awal = MathTex(
            r"\frac{4}{5}", # 0
            r"\times",      # 1
            r"15",          # 2
            r"=",           # 3
            r"30\%",        # 4
            r"\times",      # 5
            r"x"            # 6
        ).scale(2).next_to(soal_teks, DOWN, buff=1)

        # Animasi transformasi dari Teks ke Rumus
        self.play(ReplacementTransform(soal_teks, persamaan_awal))
        self.wait(1)

        # --- BAGIAN 3: MENYELESAIKAN RUAS KIRI (4/5 x 15) ---

        # Fokuskan pandangan ke ruas kiri
        fokus_kiri = SurroundingRectangle(persamaan_awal[0:3], color=YELLOW, buff=0.2)
        self.play(Create(fokus_kiri))

        # Animasi Coret: 15 dibagi 5
        garis_coret_15 = Line(persamaan_awal[2].get_corner(DL), persamaan_awal[2].get_corner(UR), color=RED)
        garis_coret_5 = Line(persamaan_awal[0][2].get_corner(DL), persamaan_awal[0][2].get_corner(UR), color=RED)

        self.play(Create(garis_coret_15), Create(garis_coret_5))

        # Hasil coretan: 15 jadi 3
        angka_3 = MathTex("3", color=YELLOW).scale(2).next_to(persamaan_awal[2], UP)
        self.play(Write(angka_3))
        self.wait(0.5)

        # Hitung hasil ruas kiri: 4 x 3 = 12
        hasil_kiri = MathTex("12").scale(2).move_to(persamaan_awal[0:3])

        # Bersihkan coretan dan ganti ruas kiri jadi "12"
        self.play(
            FadeOut(fokus_kiri),
            FadeOut(garis_coret_15), FadeOut(garis_coret_5), FadeOut(angka_3),
            ReplacementTransform(persamaan_awal[0:3], hasil_kiri)
        )
        self.wait(1)

        # --- BAGIAN 4: MEMINDAHKAN RUAS KANAN ---

        # Persamaan saat ini: 12 = 30% * x
        # Kita ubah 30% menjadi pecahan 3/10
        pecahan_30 = MathTex(r"\frac{3}{10}").scale(2).move_to(persamaan_awal[4])

        self.play(Transform(persamaan_awal[4], pecahan_30))
        self.wait(0.5)

        # SEKARANG BAGIAN KRUSIAL: Pindah Ruas (Pecahan Dibalik)
        # Kita ingin x sendirian. 3/10 pindah ke kiri jadi 10/3

        # Buat target struktur akhir: x = 12 * 10/3
        persamaan_final = MathTex(
            r"x",               # 0
            r"=",               # 1
            r"12",              # 2
            r"\times",          # 3
            r"\frac{10}{3}" # 4 (Pecahan terbalik)
        ).scale(2).move_to(persamaan_awal)

        # Animasi Pindah Ruas
        self.play(
            # x tetap/geser
            Transform(persamaan_awal[6], persamaan_final[0]),
            # = tetap
            Transform(persamaan_awal[3], persamaan_final[1]),
            # 12 geser
            Transform(hasil_kiri, persamaan_final[2]),
            # kali muncul
            Write(persamaan_final[3]),
            # 30% (3/10) pindah dan berputar jadi 10/3
            Transform(persamaan_awal[4], persamaan_final[4], path_arc=-PI/2),
            # Hapus tanda kali lama
            FadeOut(persamaan_awal[5])
        )
        self.wait(1)

        # --- BAGIAN 5: HITUNG HASIL AKHIR ---

        # Coret 12 dan 3
        garis_coret_12 = Line(persamaan_final[2].get_corner(DL), persamaan_final[2].get_corner(UR), color=RED)
        garis_coret_3_bawah = Line(persamaan_final[4][2].get_corner(DL), persamaan_final[4][2].get_corner(UR), color=RED)

        self.play(Create(garis_coret_12), Create(garis_coret_3_bawah))

        # Hasil 12/3 = 4
        angka_4 = MathTex("4", color=YELLOW).scale(2).next_to(persamaan_final[2], UP)
        self.play(TransformFromCopy(persamaan_final[2], angka_4))

        # Hasil Akhir: 4 * 10 = 40
        jawaban_akhir = MathTex("x = 40", color=YELLOW, font_size=100).next_to(persamaan_final, DOWN, buff=1)
        bg_jawaban = SurroundingRectangle(jawaban_akhir, color=YELLOW, fill_opacity=0.2)

        self.play(Write(jawaban_akhir), Create(bg_jawaban))
        self.wait(1)

        # Pilih Opsi Jawaban (B)
        opsi_b = Text("Jawaban: B (40)", weight=BOLD, color=BLACK).to_edge(DOWN, buff=2)
        bg_opsi = BackgroundRectangle(opsi_b, color=WHITE, fill_opacity=1, buff=0.2)

        self.play(FadeIn(bg_opsi), Write(opsi_b))
        self.wait(2)
```

```python
from manim import *

# --- KONFIGURASI VERTIKAL ---
config.pixel_height = 1920
config.pixel_width = 1080
config.frame_height = 16.0
config.frame_width = 9.0

%%manim -r 1080,1920 --disable_caching SolusiPersentase

class SolusiPersentase(Scene):
    def construct(self):
        # --- BAGIAN 1: MENAMPILKAN SOAL ---

        # Header
        header = Text("TRIK TIU CPNS", font_size=60, weight=BOLD, color=YELLOW).to_edge(UP, buff=1)

        # Teks Soal (Kita susun agar mudah dibaca di layar HP)
        soal_teks = VGroup(
            Text("4/5 dari 15", font_size=50),
            Text("akan sama dengan", font_size=40),
            Text("30% dari ...", font_size=50, color=BLUE)
        ).arrange(DOWN, buff=0.3).next_to(header, DOWN, buff=1)

        self.play(Write(header), FadeIn(soal_teks, shift=UP))
        self.wait(1)

        # --- BAGIAN 2: MENERJEMAHKAN KE MATEMATIKA ---

        # Kita buat persamaan matematika yang posisinya pas di bawah teks soal
        # Format: [0]=4/5, [1]=x, [2]=15, [3]=, [4]=30%, [5]=x, [6]=...
        persamaan_awal = MathTex(
            r"{{4} \over {5}}", # 0
            r"\times",          # 1
            r"{{15}}",          # 2
            r"=",               # 3
            r"{{30\%}}",        # 4
            r"\times",          # 5
            r"x"                # 6
        ).scale(2).next_to(soal_teks, DOWN, buff=1)

        # Animasi transformasi dari Teks ke Rumus
        self.play(
            ReplacementTransform(soal_teks[0], persamaan_awal[0:3]), # "4/5 dari 15" -> 4/5 x 15
            ReplacementTransform(soal_teks[1], persamaan_awal[3]),   # "sama dengan" -> =
            ReplacementTransform(soal_teks[2], persamaan_awal[4:7]), # "30% dari..." -> 30% x a
        )
        self.wait(1)

        # --- BAGIAN 3: MENYELESAIKAN RUAS KIRI (4/5 x 15) ---

        # Fokuskan pandangan ke ruas kiri
        fokus_kiri = SurroundingRectangle(persamaan_awal[0:3], color=YELLOW, buff=0.2)
        self.play(Create(fokus_kiri))

        # Animasi Coret: 15 dibagi 5
        garis_coret_15 = Line(persamaan_awal[2].get_corner(DL), persamaan_awal[2].get_corner(UR), color=RED)
        garis_coret_5 = Line(persamaan_awal[0][2].get_corner(DL), persamaan_awal[0][2].get_corner(UR), color=RED)

        self.play(Create(garis_coret_15), Create(garis_coret_5))

        # Hasil coretan: 15 jadi 3
        angka_3 = MathTex("3", color=YELLOW).scale(2).next_to(persamaan_awal[2], UP)
        self.play(Write(angka_3))
        self.wait(0.5)

        # Hitung hasil ruas kiri: 4 x 3 = 12
        hasil_kiri = MathTex("12").scale(2).move_to(persamaan_awal[0:3])

        # Bersihkan coretan dan ganti ruas kiri jadi "12"
        self.play(
            FadeOut(fokus_kiri),
            FadeOut(garis_coret_15), FadeOut(garis_coret_5), FadeOut(angka_3),
            ReplacementTransform(persamaan_awal[0:3], hasil_kiri)
        )
        self.wait(1)

        # --- BAGIAN 4: MEMINDAHKAN RUAS KANAN ---

        # Persamaan saat ini: 12 = 30% * x
        # Kita ubah 30% menjadi pecahan 3/10
        pecahan_30 = MathTex(r"{{3} \over {10}}").scale(2).move_to(persamaan_awal[4])

        self.play(Transform(persamaan_awal[4], pecahan_30))
        self.wait(0.5)

        # SEKARANG BAGIAN KRUSIAL: Pindah Ruas (Pecahan Dibalik)
        # Kita ingin x sendirian. 3/10 pindah ke kiri jadi 10/3

        # Buat target struktur akhir: x = 12 * 10/3
        persamaan_final = MathTex(
            r"x",               # 0
            r"=",               # 1
            r"12",              # 2
            r"\times",          # 3
            r"{{10} \over {3}}" # 4 (Pecahan terbalik)
        ).scale(2).move_to(persamaan_awal)

        # Animasi Pindah Ruas
        self.play(
            # x tetap/geser
            Transform(persamaan_awal[6], persamaan_final[0]),
            # = tetap
            Transform(persamaan_awal[3], persamaan_final[1]),
            # 12 geser
            Transform(hasil_kiri, persamaan_final[2]),
            # kali muncul
            Write(persamaan_final[3]),
            # 30% (3/10) pindah dan berputar jadi 10/3
            Transform(persamaan_awal[4], persamaan_final[4], path_arc=-PI/2),
            # Hapus tanda kali lama
            FadeOut(persamaan_awal[5])
        )
        self.wait(1)

        # --- BAGIAN 5: HITUNG HASIL AKHIR ---

        # Coret 12 dan 3
        garis_coret_12 = Line(persamaan_final[2].get_corner(DL), persamaan_final[2].get_corner(UR), color=RED)
        garis_coret_3_bawah = Line(persamaan_final[4][2].get_corner(DL), persamaan_final[4][2].get_corner(UR), color=RED)

        self.play(Create(garis_coret_12), Create(garis_coret_3_bawah))

        # Hasil 12/3 = 4
        angka_4 = MathTex("4", color=YELLOW).scale(2).next_to(persamaan_final[2], UP)
        self.play(TransformFromCopy(persamaan_final[2], angka_4))

        # Hasil Akhir: 4 * 10 = 40
        jawaban_akhir = MathTex("x = 40", color=YELLOW, font_size=100).next_to(persamaan_final, DOWN, buff=1)
        bg_jawaban = SurroundingRectangle(jawaban_akhir, color=YELLOW, fill_opacity=0.2)

        self.play(Write(jawaban_akhir), Create(bg_jawaban))
        self.wait(1)

        # Pilih Opsi Jawaban (B)
        opsi_b = Text("Jawaban: B (40)", weight=BOLD, color=BLACK).to_edge(DOWN, buff=2)
        bg_opsi = BackgroundRect(opsi_b, fill_color=WHITE, fill_opacity=1, buff=0.2)

        self.play(FadeIn(bg_opsi), Write(opsi_b))
        self.wait(2)
```
